```python
#1.Project Introduction 

 #"I worked on an Exploratory Data Analysis (EDA) project using the Zomato restaurant dataset to understand customer behavior and restaurant trends."
    #"Tools used; python, pandas , numpy, matplotlib , seaborn for data cleaning and visualization."

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
#Here, I loaded the Zomato Dataset using Pandas and Displayed the first few rows to understand the structure of the data.
```

```python
dataframe = pd.read_csv(r"C:\users\hp\Downloads\Zomato-data-.csv")
print(dataframe.head())
```

```python
#3. Rate Column Cleaning 
 #The rate column contained values in string format (e.g.4.1/5).I created a custom function to extract the numeric rating and converted it into a float for analysis.
```

```python
def handleRate(value):
    value=str(value).split('/')
    value=value[0];
    return float(value)

dataframe['rate']=dataframe['rate'].apply(handleRate)
print(dataframe.head())
```

```python
#Reataurant type count plot 
# A count plot was used to visualize the number of restaurants based on their listed type.
```

```python
sns.countplot(x=dataframe['listed_in(type)'])
plt.xlabel("Type of restaurant")
```

```python
#Votes Analysis by Restaurant Type 
# " In this step ,I grouped the Data by restaurant type and calculated the total number of votes recieved by each type . A line plot was used ti visualize customer engagement.
```

```python
grouped_data = dataframe.groupby('listed_in(type)')['votes'].sum()
result = pd.DataFrame({'votes': grouped_data})
plt.plot(result,c='green',marker='o')
plt.xlabel('Type of restaurant')
plt.ylabel('Votes')
```

```python
#Maximum Votes Restaurant 
#This code identifies the restaurant (s)with the maximum number of votes in the dataset.
```

```python
max_votes = dataframe['votes'].max()
restaurant_with_max_votes = dataframe.loc[dataframe['votes'] == max_votes,'name']
print('Restaurant(s) with the maximum votes:')
print(restaurant_with_max_votes)
```

```python
#Heatmap: Restaurant Type vs Online Order 
#" A heatmap was created using a pivot table to visualize the relationship between restaurant type and online ordering availability."
```

```python
pivot_table = dataframe.pivot_table(index='listed_in(type)',
columns= 'online_order',aggfunc='size',fill_value=0)
sns.heatmap(pivot_table,annot=True ,cmap='YlGnBu',fmt='d')
plt.title('Heatmap')
plt.xlabel('Online Order')
plt.ylabel('Listed In Type)')
plt.show()
```
